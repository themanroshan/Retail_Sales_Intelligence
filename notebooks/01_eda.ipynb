````python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))

import pandas as pd
import plotly.express as px
from src.config import DATA_PROCESSED
from src.analytics import kpi_summary, top_departments, top_aisles, top_products, demand_heatmap, daily_orders

aisles      = pd.read_parquet(DATA_PROCESSED / 'aisles.parquet')
departments = pd.read_parquet(DATA_PROCESSED / 'departments.parquet')
products    = pd.read_parquet(DATA_PROCESSED / 'products.parquet')
orders      = pd.read_parquet(DATA_PROCESSED / 'orders.parquet')
op          = pd.read_parquet(DATA_PROCESSED / 'order_products.parquet')

print('Shapes:')
for name, df in dict(aisles=aisles, departments=departments, products=products, orders=orders, order_products=op).items():
    print(f'  {name}: {df.shape}')
````

## Headline KPIs

````python
kpi_summary(orders, op)
````

## Top departments and aisles

````python
td = top_departments(op, products, departments, n=10)
px.bar(td, x='order_lines', y='department', orientation='h', title='Top 10 Departments')
````

````python
ta = top_aisles(op, products, aisles, n=10)
px.bar(ta, x='order_lines', y='aisle', orientation='h', title='Top 10 Aisles')
````

````python
top_products(op, products, n=15)
````

## Time patterns

````python
heat = demand_heatmap(orders)
pivot = heat.pivot(index='order_dow', columns='order_hour_of_day', values='order_count').fillna(0)
px.imshow(pivot, aspect='auto', title='Day of Week × Hour of Day')
````

````python
daily = daily_orders(orders)
px.line(daily, x='order_date', y='order_count', title='Daily Orders')
````
